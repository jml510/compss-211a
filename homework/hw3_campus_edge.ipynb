## 1. Read the API request

The data came from the City of Berkeley's API. An API request is a web address plus a set of *parameters*: settings that tell the server which records you want. In Python we store the parameters in a dictionary.

The cell below shows the exact request we used. You don't need to run it against the internet. Read it and the comments, then change it in the next cell.

One detail to know: this API sends back at most 1,000 records per request. Our data has 1,896 records, so we had to ask twice. The `$offset` parameter says how many records to skip. The first request uses `$offset` 0, and the second uses 1000 to skip the records we already have. This is called *paging*.

Look at the parameter names: `$select`, `$where`, `$order`, `$limit`. They should look familiar. This API's query language is a close relative of the SQL you wrote in Week 5: the server runs your request against the city's database and sends back the matching rows.

```python
API_ENDPOINT = "https://data.cityofberkeley.info/resource/p88g-6gs2.json"

query_params = {
    # Which columns to send back.
    "$select": "case_id, date_opened, case_status, last_action_date, date_closed, "
               "case_request, city, state, latitude, longitude",
    # Which rows to keep: cases opened in 2025, inside a box around campus.
    "$where": (
        "date_opened >= '2025-01-01' AND date_opened < '2026-01-01' "
        "AND latitude BETWEEN 37.862 AND 37.884 "
        "AND longitude BETWEEN -122.274 AND -122.244"
    ),
    # Sort by case ID so that every page comes back in the same order.
    "$order": "case_id",
    # At most 1,000 records per request.
    "$limit": 1000,
    # How many records to skip before starting.
    "$offset": 0,
}

# This is how the request would be sent. We don't call it in this homework.
def fetch_page(params):
    response = requests.get(API_ENDPOINT, params=params, timeout=30)
    response.raise_for_status()
    return response.json()

query_params
```

Your snapshot was downloaded on **August 23, 2026**. On **September 23, 2026** we ran the exact same request again and saved the result as `hw3_campus_edge_311_2025_refresh.csv`. Same query, same year, same area. In a real job, this is the moment a colleague asks: "Why don't your numbers match last month's report?"

To find out, you'll **join** the two tables on `case_id`, as in Week 5. There are three possibilities for each case:

- it is in **both** extracts;
- it is only in the **August** snapshot (it disappeared);
- it is only in the **September** refresh (it is new).

Remember what `LEFT JOIN` does: it keeps every row of the first table. When a row has no match in the second table, the second table's columns come back empty (`NULL`). So you can find the rows with no match by asking where the second table's `case_id` `IS NULL`.

```python
# Worked example: cases that are in September but were NOT in August.
# Start FROM september, so every September case is kept.
# Where August has no matching case, a.case_id is NULL.
new_cases = sql("""
    SELECT s.case_id, s.date_opened, s.case_status, s.last_action_date
    FROM september AS s
    LEFT JOIN august AS a
        ON s.case_id = a.case_id
    WHERE a.case_id IS NULL
""")
print(len(new_cases), "new case(s)")
new_cases
```
