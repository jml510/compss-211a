# SF311 requests

Data covers August 3–9, 2026 · Extracted September 22, 2026

```python
import pandas as pd

requests = pd.read_csv(
    "data/sf311_requests.csv", dtype={"request_id": "string", "district": "Int64"}
)
print("Requests:", len(requests))
```

## By district


```python
original_report = (
    requests.groupby("district")["hours_to_close"]
    .agg(requests="size", with_time="count", median_hours="median")
    .round(2)
    .reset_index()
)
original_report
```

## Data checks


```python
audit = requests.copy()
audit["recorded_hours"] = (
    pd.to_datetime(audit["closed_at"]) - pd.to_datetime(audit["opened_at"])
).dt.total_seconds() / 3600
```

## Street and Sidewalk Cleaning


```python
category_name = "Street and Sidewalk Cleaning"
selected = requests.loc[requests["category"] == category_name].copy()
print("Selected requests:", len(selected))
```

```python
draft_report = (
    selected.groupby("district", dropna=False)["hours_to_close"]
    .agg(requests="size", with_time="count", median_hours="median")
    .round(2)
    .reset_index()
)
draft_report
```

## About the data

This report summarizes 16,876 SF311 requests opened August 3–9, 2026, using records downloaded on September 22. Each input row is one request; each report row summarizes a district. A closure timestamp records when a request was marked closed in the system. It does not tell us whether the problem was fixed.

The [data reference](data/README.md) explains the columns, which records are included and how the report handles missing or inconsistent closure times.
